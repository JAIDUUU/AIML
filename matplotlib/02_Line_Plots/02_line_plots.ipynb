```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv("oscars_df.csv")
df1 = df[["Film", "Year of Release", "Audience Count"]].dropna()

df1 = df1[
    (df1["Year of Release"] >= 2014) &
    (df1["Year of Release"] <= 2020)
]

df1 = df1.loc[
    df1.groupby("Year of Release")["Audience Count"].idxmax()
]


df1 = df1.sort_values("Year of Release")

plt.plot(
    df1["Year of Release"],
    df1["Audience Count"],
    marker="o"
)

plt.xlabel("Year of Release")
plt.ylabel("Audience Count")
plt.title("Highest Audience Count by Year")
plt.show()
```
