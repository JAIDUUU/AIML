```python
import pandas as pd

df = pd.DataFrame({
    "country": ["USA", "USA", "India", "India"],
    "year": [2020, 2021, 2020, 2021],
    "sales": [100, 120, 90, 110],
    "profit": [20, 25, 18, 22]
})

melted = df.melt(
    id_vars=["country", "year"],
    value_vars=["sales", "profit"],
    var_name="metric",
    value_name="value"
)

print(melted)

pivoted = melted.pivot(
    index=["country", "year"],
    columns="metric",
    values="value"
)

print(pivoted)
```
