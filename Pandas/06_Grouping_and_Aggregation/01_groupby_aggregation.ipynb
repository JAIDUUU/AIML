```python
# df.groupby()
# We use the groupby method to split data into multiple groups so that we can apply
# some aggregation functions on it.
import pandas as pd
df=pd.read_csv("grouping_aggregation_practice.csv")
df.groupby("country")["income"].mean()

df.groupby("country")["income"].sum()
df.groupby("country")["income"].count()
df.groupby("country")["income"].max()
df.groupby("country")["income"].min()
df.groupby("country")["income"].std()

# applies multiple aggregate functions
df.groupby("country")["income"].agg(["mean","min","max"])
df.groupby("country")["income"].aggregate(["mean", "min", "max"])

# rename aggregate 

df.groupby("country")["income"].agg(avg_salary="mean", max_salary="max") 
# aggregatre on multiple cols
df.groupby("country").agg({"age":"mean","income":"mean"})

# rename aggregates on multiple cols
df.groupby("country").agg(
    avg_age=("age","mean"),
    avg_salary=("income","mean")
)

```
