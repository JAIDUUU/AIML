```python
import pandas as pd
df=pd.read_csv("raw_data_with_more_columns.csv")
df_renamed=df.rename(columns={"name":"full name"})

df_recorded=df_renamed[["country", "age", "income", "occupation"]]
df_recorded

df_filtered=df_renamed[df_renamed["age"]>20]
df_filtered
df_reset=df_filtered.reset_index(drop=True)
df_reset
```

```python
# 2.Sorting / Ranking
# Sorting - values & index 
df2=df.copy()
df2.sort_values("income")
df2.sort_values("income",ascending=False)
df2.sort_values(["age","income"])


sorted_df2=df2.sort_values(["age","income"])
sorted_df2.sort_index()

# Ranking 
df2["Ranking"]=df2["income"].rank(ascending=False,method="dense")
df2["Ranking"]=df2["income"].rank(ascending=False,method="min")
df2["Ranking"]=df2["income"].rank(ascending=False,method="max")
df2
```
