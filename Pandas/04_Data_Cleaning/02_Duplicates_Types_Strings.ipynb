```python
# • Handle Duplicate Values (NaN )
# 1. df.duplicated()- Find duplicates
# 2. df.drop_duplicates()- Remove duplicate rows

import pandas as pd 

df=pd.read_csv("raw_data.csv")
print()
df.duplicated()
df.duplicated("name")
df.duplicated(["country", "gender"]) 
df.drop_duplicates()

# • Changing Data Types
# Changing Data Types
# 1. df.astype(new_type).- Changes dtype.
# 2. pd.to_datetime() - To convert a string, number, or array-like object to a
# datetime object.
df2=df.copy()
df2=df2.fillna(0)
df2["age"]=df2["age"].astype(int)


date_str="2025-12-01"
date=pd.to_datetime(date_str)
print(date)
print(type(date))

#String Cleaning
df2["name"].str.lower()
df2["name"].str.capitalize()
df2["name"].str.upper()
df2["name"].str.strip()
df2["name"].str.split(" ")
df2["country"].str.contains("USA")
df2["country"].str.contains("india",case=False)
```
